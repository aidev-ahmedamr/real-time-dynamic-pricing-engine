```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("dynamic_pricing_raw.csv")

df.head()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df["current_price"],
    df["demand"],
    alpha=0.2
)

plt.xlabel("Current Price")
plt.ylabel("Demand")
plt.title("Price vs Demand")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df["competitor_price"],
    df["demand"],
    alpha=0.2
)

plt.xlabel("Competitor Price")
plt.ylabel("Demand")

plt.show()
```
